##### now,we will start working with pytorch

##### First we need to convert the numpy arrays into tensors to work with pytorch

```python
X_train_tensor= torch.from_numpy(X_train)
X_test_tensor= torch.from_numpy(X_test)
y_train_tensor= torch.from_numpy(y_train)
y_test_tensor= torch.from_numpy(y_test)
```

##### Define the model


```python
x.shape[1] #input features
```

```python
class SimpleNN():
   def __init__(self,x):
    self.weights= torch.rand(x.shape[1],1,dtype=torch.float64,requires_grad=True) #for 30inputs ,initialize 30 wieghtswitht value 1
    self.bias= torch.zeros(1,dtype=torch.float64,requires_grad=True)


   def forward_pass(self,x):
    z=torch.matmul(x,self.weights)+self.bias
    y_pred= torch.sigmoid(z)
    return y_pred

   def loss_function(self,y_pred,y):
    epsilon=1e-7
    y_pred=torch.clamp(y_pred,epsilon,1-epsilon)

    #calculate loss
    loss=-(y*torch.log(y_pred)+(1-y)*torch.log(1-y_pred)).mean()
    return loss


```

#### Define NN parameters

```python
learning_rate=0.1
epochs=25
```

##### Training pipeline

```python
#create the model
model= SimpleNN(X_train_tensor)
### to train the model
# first: forward pass, calculate loss, backward pass, and parameter update
# we do these 4 things in a loop

for epoch in range(epochs):
  #forward pass
  y_pred=model.forward_pass(X_train_tensor)

  #calculate loss
  loss=model.loss_function(y_pred,y_train_tensor)

  #backward pass
  loss.backward()

  #update parameters
  with torch.no_grad():
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad

  #zero gradients
  model.weights.grad.zero_()
  model.bias.grad.zero_()

  #print loss in each epoch
  print(f'epoch:{epoch+1}, loss:{loss.item()}')
```

#####  we can see that in each epoch,loss is  reducing which is good
