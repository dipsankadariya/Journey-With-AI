# 🧠 PyTorch `nn` Module (Quick Notes)

## What is `torch.nn`?
- Core module for building neural networks
- Provides layers, activations, losses, and containers

## Main Components
- `nn.Module` → Base class for all models
- `nn.Linear` → Fully connected layer
- `nn.Conv2d` → Convolution layer
- `nn.LSTM`, `nn.GRU` → Recurrent layers
- `nn.Embedding` → Word embeddings

## Activation Functions
- `nn.ReLU`
- `nn.Sigmoid`
- `nn.Tanh`
- `nn.Softmax`

## Loss Functions
- `nn.MSELoss`
- `nn.CrossEntropyLoss`
- `nn.BCELoss`

## Model Flow
- Define layers in `__init__`
- Define computation in `forward()`
- PyTorch handles backprop automatically

## Common Utilities
- `nn.Sequential` → Stack layers easily
- `model.parameters()` → Get trainable params
- `model.train()` / `model.eval()`

## Typical Usage
- Forward pass → Loss → Backward → Optimizer step


##### We will create a neural network that takes 5 inputs and produces a output(binary classification )


```python
#### Create a model class
import torch
import torch.nn as nn

class Model(nn.Module):

  def __init__(self,input_features):
    super().__init__()
    self.linear= nn.Linear(input_features,1)
    self.sigmoid=nn.Sigmoid()

  def forward(self,features):
    output=self.linear(features)
    output=self.sigmoid(output)
    return output
```

```python
# Create a simple dataset
features = torch.rand(10, 5)  # 10 rows, 5 columns

# Create the model
model = Model(features.shape[1])

# Call the model
output = model(features)
print(output)
```

```python
#### showing weights
model.linear.weight
```

```python
#### showing bias
model.linear.bias
```

#### Let's build a bigger neural network
- 5 input features
- 1 hidden layer with 3 nodes
- 1 output node
for hidden we will use relu and for output we use sigmoid


```python
class Model(nn.Module):
  def __init__(self,input_features):
    super().__init__()
    self.linear1=nn.Linear(input_features,3)
    self.relu=nn.ReLU(3)
    self.linear2=nn.Linear(3,1)
    self.sigmoid=nn.Sigmoid()

  def forward(self,df):
    output=self.linear1(df)
    output=self.relu(output)
    output=self.linear2(output)
    output=self.sigmoid(output)
    return output
```

```python
df=torch.rand(10,5)
model=Model(df.shape[1])
output=model(df)
print(output)
```

```python
####Visualizing the nn
from torchinfo import summary
summary(model,input_size=(10,5))
```

##### the above is one way to do it, to do it even better, we can use containers like squential/


```python
class Model(nn.Module):

  def __init__(self,input_features):
    super().__init__()
    self.neural_network=nn.Sequential(
    nn.Linear(input_features,3),
    nn.ReLU(),
    nn.Linear(3,1),
    nn.Sigmoid()
    )

  def forward(self,df):
    output=self.neural_network(df)
    return output
```

```python
df=torch.rand(10,5)
model=Model(df.shape[1])
output=model(df)
print(output)
```

```python
####Visualizing the nn
from torchinfo import summary
summary(model,input_size=(10,5))
```
