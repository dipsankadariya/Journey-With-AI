## Matrix operation

```python
a=torch.randint(size=(2,3),low=0,high=10)
b=torch.randint(size=(3,2),low=0,high=10)
print(a)
print(b)
```

```python
##matrix mul
torch.matmul(a,b)
```

```python
#dot product
vector1= torch.tensor([1,2])
vector2=torch.tensor([3,4])

torch.dot(vector1,vector2)
```

```python
##transpose
print(a)
torch.transpose(a,0,1)
```

```python
#determinant
g=torch.randint(size=(3,3),low=0,high=10,dtype=torch.float32)
print(g)
torch.det(g)
```

```python
#inverse
torch.inverse(g)
```

## Inplace operations

```python
#in-place operation modifies the tensor itself instead of creating a new one.
m=torch.rand(2,3)
n=torch.rand(2,3)
print(m)
print(n)
```

```python
m.add_(n)
```

```python
m #it modified m instead of creating a new tensor, that would occupy the space, tosolve  this problem we used inplace operations
```

```python
n
```

```python
torch.relu(m) #this will craete a completely new tensor
```

```python
m.relu_()#this wll modfy exisiting ones
```

```python
m
```
