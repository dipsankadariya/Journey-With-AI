####  Stemming->bringing  the words their root form by removing the suffixes and prefixes

##### popular stemmng alg->porter stemmer,snowball stemmer

```python
import pandas as pd
import numpy as np
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps=PorterStemmer()
def stem_words(text):
  return " ".join([ps.stem(word) for word in text.split()])
```

```python
txt="run runs running runned"
stem_words(txt)
```

```python
sent="The students were studying different subjects and working on their assignments."
stem_words(sent)
```

##### Problem with over-stemming
##### It may reduce different words to the same root, even when meanings differ.
Example:
universe → univers
university → univers

##### Problem with under-stemming
##### Sometimes it fails to reduce similar words to the same stem.
Example:
analysis → analysi
analyze → analyz

##### Problem with non-real words
##### Stems may not be valid English words.
Example:
studies → studi
relational → relat

##### Problem with ignoring context
##### It does not understand grammar or meaning.
Example:
better → better (not linked to "good")

##### Problem with English-focus
##### Mainly designed for English; less effective for other languages.
Example:
खेल्नु (Nepali "to play") → may not stem correctly

##### Lemmatization-> unlike stemming, reduces the inflected words properly ensuring that root  word belongs to the language. In lemmatization root word is called lemma. A lemma is the canonical form, dictionary form, or citation form of a set of words

```python
from string import punctuation
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

nltk.download('punkt')
nltk.download('wordnet')

wordnet_lemmatizer = WordNetLemmatizer()

sentence = "Wow! The quick brown foxes were jumping over lazy dogs; they laughed, barked, and ran away."

# Tokenize
tokens = nltk.word_tokenize(sentence)
print("Tokens before removing punctuations:", tokens)

# Remove punctuations
tokens = [word for word in tokens if word not in punctuation]
print("Tokens after removing punctuations:", tokens)

# Lemmatize as VERB
for word in tokens:
    lemma = wordnet_lemmatizer.lemmatize(word, pos=wordnet.VERB)
    print("{:20}{:20}".format(word, lemma))
```

#### Simple example of text-preprocessing on sample dataset

```python
import requests
import pandas as pd
import numpy as np
```

```python
url = "https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page=1"
```

```python
# fetch genre details from url
movie_genre_url = "https://api.themoviedb.org/3/genre/movie/list?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US"
response = requests.get(movie_genre_url)
movies_genre_details = response.json()["genres"]
```

```python
movies_genre_details
```

```python
#mapping id -> name
genre = {}
for gen in movies_genre_details:
    genre[gen["id"]] = gen["name"]
```

```python
all_movies = []
```

```python

# loop over pages
for page in range(1, 472):
    print(f"Fetching page {page}...")

    movies_url = f"https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page={page}"
    response = requests.get(movies_url)
    data = response.json()

    # loop through each movie in this page inside the page loop
    for movie in data["results"]:
        # map genre_ids to genre names
        movies_genre_names = []
        for gen_id in movie["genre_ids"]:
            if gen_id in genre:
                movies_genre_names.append(genre[gen_id])
            else:
                movies_genre_names.append("Unknown")

        # append to all_movies
        movie_data = {
            "name": movie["title"],
            "overview": movie["overview"],
            "genre": movies_genre_names
        }
        all_movies.append(movie_data)
```

```python
print(all_movies[:5])
```

```python
df=pd.DataFrame(all_movies)
```
