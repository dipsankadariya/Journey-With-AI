#### removing stopwords

```python
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
def remove_stopwords(text):
    new_text = []
    for word in text.split():
        if word.lower() not in stopwords.words('english'):
            new_text.append(word)
    return ' '.join(new_text)
```

```python
sample_text="She said that she would call me when she arrives at the office, but I haven’t heard from her yet."
```

```python
remove_stopwords(sample_text)
```

#### Handling emojis

###### we have two options->either remove the emojis or replace them with their textual meaning

```python
import re

def remove_emojis(text):

    emoji_pattern = re.compile(
        "["
        u"\U0001F600-\U0001F64F"  # Emoticons
        u"\U0001F300-\U0001F5FF"  # Symbols & Pictographs
        u"\U0001F680-\U0001F6FF"  # Transport & Map Symbols
        u"\U0001F1E0-\U0001F1FF"  # Flags
        u"\U00002700-\U000027BF"  # Dingbats
        u"\U0001F900-\U0001F9FF"  # Supplemental Symbols & Pictographs
        u"\U00002600-\U000026FF"  # Misc symbols
        "]+", flags=re.UNICODE
    )
    return emoji_pattern.sub(r'', text)
```

```python
text = "Hallo 😊! i love Python 🐍🚀"
remove_emojis(text)
```

```python
remove_emojis("lmaoooooooooooooooo😂🤣")
```

```python
# sentence tokenizations
txt="I am from Birtamode. I live near lfes school. blablaabla"
txt.split('.')
```

```python
# problem with split
txt = "I live in the U.S. I love it. The number is 3.5. Dr. Bob is here."

# Using simple split('.') method
sentences_split = txt.split('.')
print(sentences_split)
```

txt.split('.'), Python cuts the text at every dot without thinking:

"U.S." → becomes 'U' and 'S' (abbreviation broken)

"3.5" → becomes '3' and '5' (number broken)

"Dr. Bob" → becomes 'Dr' and 'Bob is here' (title broken)
