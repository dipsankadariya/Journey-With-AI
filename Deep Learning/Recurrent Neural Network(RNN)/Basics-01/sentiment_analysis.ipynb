#### load the dataset

```python
(X_train,y_train),(X_test,y_test)=imdb.load_data()
```

```python
from keras.utils import pad_sequences
```

##### we  use pad_sequences to make the sentences of equal length
###### padding="post" adds padding at end of sentence
###### maxlen=100, we are keeping maxlen=100 for now so that training becomes faster.(optional)

```python
X_train = pad_sequences(X_train, padding="post",maxlen=100)
X_test  = pad_sequences(X_test, padding="post",maxlen=100)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
```

#### model architecture

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

model = Sequential()
model.add(Embedding(input_dim=10000, output_dim=2, input_shape=(100,)))
model.add(SimpleRNN(32, return_sequences=False))
model.add(Dense(1, activation="sigmoid"))

model.summary()
```

#### complie and train

```python
model.compile(loss="binary_crossentropy", optimizer="adam", metrics=["accuracy"])
```

```python
history=model.fit(X_train, y_train, epochs=5, validation_data=(X_test, y_test))
```
