```python
import numpy as np
data = [
    "The rain started falling just as the market began to close.",
    "She found an old letter hidden inside the wooden drawer.",
    "Artificial intelligence is changing the way people learn new skills.",
    "A small cat was sleeping under the bright yellow car.",
    "He decided to wake up early and watch the sunrise alone.",
    "The computer screen flickered before suddenly turning black.",
    "They built a tiny house near the edge of the forest.",
    "Music played softly in the background during the long drive.",
    "The coffee tasted better on cold winter mornings.",
    "She laughed when she realized the keys were in her pocket."
]
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer

tokenizer = Tokenizer(oov_token="unknown")
```

```python
tokenizer.fit_on_texts(data)
```

```python
tokenizer.word_index
```

```python
tokenizer.document_count
```

```python
sequences=tokenizer.texts_to_sequences(data)
```

```python
sequences
```

```python
from keras.utils import pad_sequences

sequences=pad_sequences(sequences,padding="post")
```
